```python
from pathlib import Path
import os
import sys
import json
import joblib
import numpy as np
import pandas as pd

# Keep matplotlib cache files inside the project.
os.environ.setdefault("MPLCONFIGDIR", str(Path.cwd() / "reports" / ".mplconfig"))

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, average_precision_score, balanced_accuracy_score
from sklearn.metrics import confusion_matrix, f1_score, matthews_corrcoef, precision_score, recall_score, roc_auc_score

# Resolve paths whether the notebook is run from the repo root or brainroute_ml_validation/.
cwd = Path.cwd().resolve()
if cwd.name == "brainroute_ml_validation":
    VALIDATION_DIR = cwd
    REPO_ROOT = cwd.parent
elif (cwd / "brainroute_ml_validation").exists():
    REPO_ROOT = cwd
    VALIDATION_DIR = cwd / "brainroute_ml_validation"
else:
    raise RuntimeError("Run this notebook from the repo root or brainroute_ml_validation folder.")

REPORTS_DIR = VALIDATION_DIR / "reports"
FIGURES_DIR = REPORTS_DIR / "figures"
REPORTS_DIR.mkdir(parents=True, exist_ok=True)
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
if str(REPO_ROOT / "scripts" / "webapp") not in sys.path:
    sys.path.insert(0, str(REPO_ROOT / "scripts" / "webapp"))

from brainroute_ml_validation.src.chemistry import standardize_smiles

sns.set_theme(style="whitegrid", context="notebook")
BLUE = "#1f77b4"
LIGHT_BLUE = "#8ecae6"
DARK_BLUE = "#0b3d91"
```

```python
# df2 is the PAMPA table prepared above. The model step also needs SMILES.
# If df2 does not include SMILES, copy it from df. If the file has missing SMILES,
# optionally try PubChem lookup by chemical_name.
pampa = df2.copy()
if "smiles" not in pampa.columns:
    pampa["smiles"] = df["smiles"] if "smiles" in df.columns else pd.NA
if "casrn" not in pampa.columns and "casrn" in df.columns:
    pampa["casrn"] = df["casrn"]

LOOKUP_MISSING_SMILES_WITH_PUBCHEM = True

def lookup_smiles(identifier):
    """Return a SMILES string from PubChem if available; otherwise return None."""
    if pd.isna(identifier) or str(identifier).strip() == "":
        return None
    try:
        import pubchempy as pcp
        hits = pcp.get_compounds(str(identifier), "name")
        if not hits:
            return None
        return hits[0].isomeric_smiles or hits[0].canonical_smiles
    except Exception:
        return None

missing_smiles = pampa["smiles"].isna() | (pampa["smiles"].astype(str).str.strip() == "")
if LOOKUP_MISSING_SMILES_WITH_PUBCHEM and missing_smiles.any():
    resolved = []
    for _, row in pampa.loc[missing_smiles].iterrows():
        smiles = lookup_smiles(row.get("casrn"))
        if smiles is None:
            smiles = lookup_smiles(row.get("chemical_name"))
        resolved.append(smiles)
    pampa.loc[missing_smiles, "smiles"] = resolved

print(f"PAMPA rows: {len(pampa)}")
print(f"Rows with usable SMILES after optional lookup: {pampa['smiles'].notna().sum()}")
pampa.head()
```

```python
from rdkit import Chem

def standardize_or_mark(smiles):
    """Use the same standardize_smiles helper used by the training pipeline."""
    if pd.isna(smiles) or str(smiles).strip() == "":
        return {"canonical_smiles": None, "inchikey": None, "murcko_scaffold": None, "valid_smiles": False, "standardization_error": "missing_smiles"}
    std = standardize_smiles(str(smiles))
    return {
        "canonical_smiles": std.canonical_smiles,
        "inchikey": std.inchikey,
        "murcko_scaffold": std.murcko_scaffold,
        "valid_smiles": std.valid,
        "standardization_error": std.error,
    }

def largest_fragment_parent(smiles):
    """Simple salt/mixture parent: keep the fragment with the most heavy atoms."""
    if pd.isna(smiles) or str(smiles).strip() == "":
        return None, None
    mol = Chem.MolFromSmiles(str(smiles))
    if mol is None:
        return None, None
    frags = Chem.GetMolFrags(mol, asMols=True, sanitizeFrags=True)
    if not frags:
        return None, None
    parent = max(frags, key=lambda m: m.GetNumHeavyAtoms())
    parent_smiles = Chem.MolToSmiles(parent, canonical=True, isomericSmiles=True)
    parent_inchikey = Chem.MolToInchiKey(parent)
    return parent_smiles, parent_inchikey

std_df = pd.DataFrame([standardize_or_mark(s) for s in pampa["smiles"]])
pampa_std = pd.concat([pampa.reset_index(drop=True), std_df], axis=1)
parent_values = pampa_std["canonical_smiles"].apply(lambda s: pd.Series(largest_fragment_parent(s), index=["parent_canonical_smiles", "parent_inchikey"]))
pampa_std = pd.concat([pampa_std, parent_values], axis=1)

invalid_pampa = pampa_std.loc[~pampa_std["valid_smiles"]].copy()
invalid_pampa.to_csv(REPORTS_DIR / "pampa_bbb_invalid_smiles.csv", index=False)

pampa_valid = pampa_std.loc[pampa_std["valid_smiles"]].copy()
pampa_duplicates = pampa_valid.loc[pampa_valid.duplicated("inchikey", keep=False)].copy()
pampa_duplicates.to_csv(REPORTS_DIR / "pampa_bbb_internal_duplicate_inchikeys.csv", index=False)
pampa_unique = pampa_valid.drop_duplicates("inchikey", keep="first").copy()

training = pd.read_csv(VALIDATION_DIR / "data" / "processed" / "standardized_molecules.csv")
train_canonical = set(training["canonical_smiles"].dropna().astype(str))
train_inchikey = set(training["inchikey"].dropna().astype(str))
train_parent = training["canonical_smiles"].dropna().astype(str).apply(lambda s: pd.Series(largest_fragment_parent(s), index=["parent_canonical_smiles", "parent_inchikey"]))
train_parent_smiles = set(train_parent["parent_canonical_smiles"].dropna().astype(str))
train_parent_inchikey = set(train_parent["parent_inchikey"].dropna().astype(str))

def overlap_reason(row):
    reasons = []
    if row["inchikey"] in train_inchikey:
        reasons.append("exact_inchikey")
    if row["canonical_smiles"] in train_canonical:
        reasons.append("canonical_smiles")
    if row["parent_inchikey"] in train_parent_inchikey:
        reasons.append("parent_inchikey")
    if row["parent_canonical_smiles"] in train_parent_smiles:
        reasons.append("parent_canonical_smiles")
    return ";".join(sorted(set(reasons)))

pampa_unique["training_overlap_reason"] = pampa_unique.apply(overlap_reason, axis=1)
overlaps_removed = pampa_unique.loc[pampa_unique["training_overlap_reason"] != ""].copy()
pampa_external = pampa_unique.loc[pampa_unique["training_overlap_reason"] == ""].copy()

overlaps_removed.to_csv(REPORTS_DIR / "pampa_bbb_training_overlaps_removed.csv", index=False)
pampa_external.to_csv(REPORTS_DIR / "pampa_bbb_external_cleaned_for_prediction.csv", index=False)

accounting = pd.DataFrame([
    {"step": "starting_pampa_rows", "n": len(pampa)},
    {"step": "missing_or_invalid_smiles_removed", "n": len(invalid_pampa)},
    {"step": "internal_duplicate_inchikey_rows_flagged", "n": len(pampa_duplicates)},
    {"step": "unique_valid_pampa_molecules", "n": len(pampa_unique)},
    {"step": "training_exact_or_parent_overlaps_removed", "n": len(overlaps_removed)},
    {"step": "final_external_molecules_for_prediction", "n": len(pampa_external)},
])
accounting.to_csv(REPORTS_DIR / "pampa_bbb_external_accounting.csv", index=False)

display(accounting)
if pampa_external.empty:
    raise ValueError("No PAMPA molecules remain for prediction after SMILES standardization and training-overlap removal.")
```

```python
# Load the same three deployed models used by the Streamlit app.
MODEL_SPECS = {
    "PaDEL+Morgan LGBM": {
        "path": VALIDATION_DIR / "models" / "padel_morgan__lightgbm__duplicate_aware_seed5.joblib",
        "feature_view": "padel_morgan",
    },
    "PaDEL+Morgan Extra Trees": {
        "path": VALIDATION_DIR / "models" / "padel_morgan__extra_trees__duplicate_aware_seed5.joblib",
        "feature_view": "padel_morgan",
    },
    "PaDEL+Morgan+Embeddings XGBoost": {
        "path": VALIDATION_DIR / "models" / "padel_morgan_embeddings__xgboost__scaffold_cv_fold1.joblib",
        "feature_view": "padel_morgan_embeddings",
    },
}

models = {name: joblib.load(spec["path"]) for name, spec in MODEL_SPECS.items()}
models["model_feature_views"] = {name: spec["feature_view"] for name, spec in MODEL_SPECS.items()}

from prediction import _build_webapp_feature_views, _predict_from_feature_views

# Use canonical SMILES so feature generation starts from the same standardized representation.
smiles_for_prediction = pampa_external["canonical_smiles"].tolist()
feature_views, failed_indices, embedding_error = _build_webapp_feature_views(smiles_for_prediction)
if embedding_error:
    print("Embedding feature warning:", embedding_error)

prediction_rows = []
model_slugs = {
    "PaDEL+Morgan LGBM": "padel_morgan_lgbm",
    "PaDEL+Morgan Extra Trees": "padel_morgan_extra_trees",
    "PaDEL+Morgan+Embeddings XGBoost": "padel_morgan_embeddings_xgboost",
}

for row_position, (_, mol_row) in enumerate(pampa_external.reset_index(drop=True).iterrows()):
    out = mol_row.to_dict()
    out["prediction_status"] = "ok"
    if row_position in failed_indices:
        out["prediction_status"] = "padel_feature_failed"
        prediction_rows.append(out)
        continue

    result, error = _predict_from_feature_views(feature_views, row_position, models)
    if error:
        out["prediction_status"] = json.dumps(error)
        prediction_rows.append(out)
        continue

    for model_name, slug in model_slugs.items():
        pred_value = result["predictions"].get(model_name)
        prob_value = result["positive_probabilities"].get(model_name)
        out[f"{slug}_prediction"] = pred_value
        out[f"{slug}_prediction_label"] = "BBB+" if pred_value == 1 else "BBB-" if pred_value == 0 else np.nan
        out[f"{slug}_bbb_positive_probability"] = prob_value

    out["consensus_prediction"] = 1 if result["ensemble_pred"] == "BBB+" else 0
    out["consensus_prediction_label"] = result["ensemble_pred"]
    out["consensus_bbb_positive_probability"] = np.mean(list(result["positive_probabilities"].values()))
    out["consensus_confidence"] = result["avg_confidence"]
    out["model_agreement_percent"] = result["agreement"]
    out["model_uncertainty_percent"] = result["uncertainty"]
    out["model_errors"] = json.dumps(result.get("model_errors", {}))
    prediction_rows.append(out)

pampa_predictions = pd.DataFrame(prediction_rows)
pampa_predictions.to_csv(REPORTS_DIR / "pampa_bbb_model_predictions.csv", index=False)
display(pampa_predictions.head())
```

```python
# Compare model predictions against the PAMPA-BBB label.
def normalize_binary_label(value):
    if pd.isna(value):
        return np.nan
    if value in [0, 1]:
        return int(value)
    text = str(value).strip().lower()
    if text in {"1", "bbb+", "positive", "fast_permeable", "permeable", "yes"}:
        return 1
    if text in {"0", "bbb-", "negative", "slow_permeable", "impermeable", "no"}:
        return 0
    return np.nan

pampa_predictions["pampa_binary"] = pampa_predictions["pampa_label"].apply(normalize_binary_label)
ok = pampa_predictions.loc[pampa_predictions["prediction_status"] == "ok"].copy()
ok = ok.loc[ok["pampa_binary"].isin([0, 1])].copy()

def metric_row(display_name, pred_col, prob_col):
    valid = ok[["pampa_binary", pred_col, prob_col]].dropna(subset=["pampa_binary", pred_col]).copy()
    y_true = valid["pampa_binary"].astype(int).to_numpy()
    y_pred = valid[pred_col].astype(int).to_numpy()
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()
    row = {
        "model": display_name,
        "n": len(valid),
        "accuracy": accuracy_score(y_true, y_pred),
        "balanced_accuracy": balanced_accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred, zero_division=0),
        "recall_sensitivity": recall_score(y_true, y_pred, zero_division=0),
        "specificity": tn / (tn + fp) if (tn + fp) else np.nan,
        "f1": f1_score(y_true, y_pred, zero_division=0),
        "mcc": matthews_corrcoef(y_true, y_pred) if len(set(y_true)) > 1 else np.nan,
        "tn": tn,
        "fp": fp,
        "fn": fn,
        "tp": tp,
    }
    valid_prob = valid.dropna(subset=[prob_col])
    if len(valid_prob) and valid_prob["pampa_binary"].nunique() == 2:
        y_prob_true = valid_prob["pampa_binary"].astype(int).to_numpy()
        y_score = valid_prob[prob_col].astype(float).to_numpy()
        row["roc_auc"] = roc_auc_score(y_prob_true, y_score)
        row["auprc"] = average_precision_score(y_prob_true, y_score)
    else:
        row["roc_auc"] = np.nan
        row["auprc"] = np.nan
    return row

metric_specs = [
    ("PaDEL+Morgan LGBM", "padel_morgan_lgbm_prediction", "padel_morgan_lgbm_bbb_positive_probability"),
    ("PaDEL+Morgan Extra Trees", "padel_morgan_extra_trees_prediction", "padel_morgan_extra_trees_bbb_positive_probability"),
    ("PaDEL+Morgan+Embeddings XGBoost", "padel_morgan_embeddings_xgboost_prediction", "padel_morgan_embeddings_xgboost_bbb_positive_probability"),
    ("Consensus", "consensus_prediction", "consensus_bbb_positive_probability"),
]
metrics = pd.DataFrame([metric_row(*spec) for spec in metric_specs if spec[1] in ok.columns])
metrics.to_csv(REPORTS_DIR / "pampa_bbb_model_comparison_metrics.csv", index=False)
display(metrics.sort_values("balanced_accuracy", ascending=False))
```

```python
# Figure 1: core PAMPA comparison metrics.
plot_metrics = metrics.melt(
    id_vars="model",
    value_vars=["accuracy", "balanced_accuracy", "roc_auc", "auprc"],
    var_name="metric",
    value_name="value",
).dropna()
plt.figure(figsize=(11, 5.5), dpi=180)
sns.barplot(data=plot_metrics, x="metric", y="value", hue="model", palette="Blues")
plt.ylim(0, 1.02)
plt.xlabel("")
plt.ylabel("Score")
plt.title("PAMPA-BBB External Comparison Metrics")
plt.legend(title="", bbox_to_anchor=(1.02, 1), loc="upper left", frameon=True)
plt.tight_layout()
plt.savefig(FIGURES_DIR / "pampa_bbb_model_metric_bars.png", bbox_inches="tight")
plt.show()

# Figure 2: confusion matrices for each model and the consensus.
fig, axes = plt.subplots(1, len(metric_specs), figsize=(4.2 * len(metric_specs), 4), dpi=180)
for ax, (display_name, pred_col, _) in zip(axes, metric_specs):
    if pred_col not in ok.columns:
        ax.axis("off")
        continue
    valid = ok[["pampa_binary", pred_col]].dropna()
    cm = confusion_matrix(valid["pampa_binary"].astype(int), valid[pred_col].astype(int), labels=[0, 1])
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False, ax=ax, square=True,
                xticklabels=["Pred BBB-", "Pred BBB+"], yticklabels=["PAMPA BBB-", "PAMPA BBB+"])
    ax.set_title(display_name)
    ax.set_xlabel("")
    ax.set_ylabel("")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "pampa_bbb_confusion_matrices.png", bbox_inches="tight")
plt.show()

# Figure 3: predicted BBB+ probability separated by the PAMPA label.
prob_cols = [(name, prob_col) for name, _, prob_col in metric_specs if prob_col in ok.columns]
prob_long = []
for name, prob_col in prob_cols:
    tmp = ok[["chemical_name", "pampa_binary", prob_col]].dropna().copy()
    tmp["model"] = name
    tmp["bbb_positive_probability"] = tmp[prob_col].astype(float)
    tmp["PAMPA label"] = tmp["pampa_binary"].map({0: "PAMPA BBB-", 1: "PAMPA BBB+"})
    prob_long.append(tmp[["chemical_name", "model", "PAMPA label", "bbb_positive_probability"]])
prob_long = pd.concat(prob_long, ignore_index=True)

plt.figure(figsize=(12, 5.5), dpi=180)
sns.boxplot(data=prob_long, x="model", y="bbb_positive_probability", hue="PAMPA label", palette=[LIGHT_BLUE, DARK_BLUE], showfliers=False)
sns.stripplot(data=prob_long, x="model", y="bbb_positive_probability", hue="PAMPA label", dodge=True, palette=[LIGHT_BLUE, DARK_BLUE], alpha=0.45, size=3, legend=False)
plt.axhline(0.5, color="black", linestyle="--", linewidth=1)
plt.ylim(0, 1.02)
plt.xlabel("")
plt.ylabel("Predicted BBB+ probability")
plt.title("Predicted BBB+ Probability by PAMPA-BBB Label")
plt.xticks(rotation=15, ha="right")
plt.legend(title="", loc="upper left")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "pampa_bbb_probability_by_pampa_label.png", bbox_inches="tight")
plt.show()

# Figure 4: model-vs-PAMPA agreement counts.
agreement_rows = []
for name, pred_col, _ in metric_specs:
    if pred_col not in ok.columns:
        continue
    valid = ok[["pampa_binary", pred_col]].dropna().copy()
    valid["model"] = name
    valid["agreement"] = np.where(valid["pampa_binary"].astype(int) == valid[pred_col].astype(int), "Agrees", "Disagrees")
    agreement_rows.append(valid[["model", "agreement"]])
agreement_long = pd.concat(agreement_rows, ignore_index=True)

plt.figure(figsize=(10, 5), dpi=180)
sns.countplot(data=agreement_long, x="model", hue="agreement", palette=[DARK_BLUE, "#d95f5f"])
plt.xlabel("")
plt.ylabel("Number of molecules")
plt.title("Agreement Between Model Predictions and PAMPA-BBB Labels")
plt.xticks(rotation=15, ha="right")
plt.legend(title="")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "pampa_bbb_prediction_agreement_counts.png", bbox_inches="tight")
plt.show()
```
