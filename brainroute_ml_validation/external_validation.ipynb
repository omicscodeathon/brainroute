```python
# Compare external molecules against the deduplicated training set by InChIKey
from rdkit import Chem
from rdkit.Chem import inchi

# This file is produced by scripts/01_standardize_and_audit.py after invalid rows,
# conflicting-label InChIKeys, and duplicate InChIKeys are removed.
deduped_training_path = Path("data/processed/standardized_molecules.csv")
df_training_unique = pd.read_csv(deduped_training_path)
training_inchikeys = set(df_training_unique["inchikey"].dropna().astype(str))

def external_row_to_inchikey(row):
    inchi_value = row.get("Inchi")
    if pd.notna(inchi_value) and str(inchi_value).strip():
        try:
            key = inchi.InchiToInchiKey(str(inchi_value).strip())
            if key:
                return key
        except Exception:
            pass

    smiles_value = row.get("SMILES")
    if pd.notna(smiles_value) and str(smiles_value).strip():
        mol = Chem.MolFromSmiles(str(smiles_value).strip())
        if mol is not None:
            return Chem.MolToInchiKey(mol)

    return None

df_external = df_external.copy()
df_external["external_inchikey"] = df_external.apply(external_row_to_inchikey, axis=1)
external_training_overlap = df_external[df_external["external_inchikey"].isin(training_inchikeys)].copy()

print(f"Deduplicated training molecules: {len(training_inchikeys):,}")
print(f"External rows: {len(df_external):,}")
print(f"External rows with InChIKeys: {df_external.external_inchikey.notna().sum():,}")
print(f"Unique external InChIKeys: {df_external.external_inchikey.nunique():,}")
print(f"External rows present in training set: {len(external_training_overlap):,}")
print(f"Unique external molecules present in training set: {external_training_overlap.external_inchikey.nunique():,}")

external_training_overlap[["compound_name", "SMILES", "external_inchikey", "BBB+/BBB-"]]

```

```python
import pandas as pd

external_data_two = "/Users/soham/Downloads/external_dataset_qsar.xlsx"
df_external_two = pd.read_excel(external_data_two, header=2)
df_external_two
```

```python
from rdkit import Chem

# Keep only the columns needed for this overlap check.
df_external_two_short = df_external_two[["SMILES", "CAS", "Name", "Species", "Activity score"]].copy()
df_external_two_short = df_external_two_short.rename(columns={
    "SMILES": "smiles",
    "CAS": "cas",
    "Name": "name",
    "Species": "species",
    "Activity score": "BBB label",
})

# Convert external SMILES to InChIKeys.
def smiles_to_inchikey(smiles):
    if pd.isna(smiles):
        return None
    mol = Chem.MolFromSmiles(str(smiles))
    if mol is None:
        return None
    return Chem.MolToInchiKey(mol)

df_external_two_short["inchikey"] = df_external_two_short["smiles"].apply(smiles_to_inchikey)

# Compare against the deduplicated training set from the training pipeline.
df_training_unique = pd.read_csv("data/processed/standardized_molecules.csv")
training_inchikeys = set(df_training_unique["inchikey"].dropna().astype(str))

df_external_two_short["present_in_training"] = df_external_two_short["inchikey"].isin(training_inchikeys)

valid_inchikeys = df_external_two_short["inchikey"].notna()
print("External rows:", len(df_external_two_short))
print("Rows with valid InChIKey:", valid_inchikeys.sum())
print("Overlaps with training:", df_external_two_short["present_in_training"].sum())
print("Not present in training:", ((~df_external_two_short["present_in_training"]) & valid_inchikeys).sum())
print("Could not convert SMILES:", (~valid_inchikeys).sum())

df_external_two_short
```
